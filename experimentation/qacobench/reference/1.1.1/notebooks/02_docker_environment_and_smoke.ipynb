```python
from pathlib import Path
import csv, json, collections
ROOT = Path.cwd()
ART = ROOT / ".artifacts/qacobench"
```

```python
preflight = ART / "preflight.json"
print(json.dumps(json.loads(preflight.read_text()), indent=2) if preflight.exists() else "Preflight result not delivered")
smoke = ART / "smoke.jsonl"
counts = collections.Counter()
if smoke.exists():
    for line in smoke.open(encoding="utf-8"):
        counts[json.loads(line)["status"]] += 1
print("Pilot outcomes:", dict(counts))
```
