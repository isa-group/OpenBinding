# Reference campaign

```python
from pathlib import Path
import csv, json, collections
ROOT = Path.cwd()
ART = ROOT / ".artifacts/qacobench"
```

```python
results = ART / "results.jsonl"
counts = collections.Counter()
triples = set()
for line in results.open(encoding="utf-8"):
    if not line.strip():
        continue
    row = json.loads(line)
    counts[row["status"]] += 1
    triples.add((row["suite"], row["instance_id"], row["engine"]))
print("Runs:", sum(counts.values()), "Unique (suite, instance, engine):", len(triples), "Expected:", 46350, "(reference campaign, one repetition)")
reasons = collections.Counter()
for line in results.open(encoding="utf-8"):
    if line.strip():
        row = json.loads(line); reasons[(row["engine"], row.get("termination_reason"))] += 1
print("Termination reasons:", dict(reasons))
print("Statuses:", dict(counts))
```
