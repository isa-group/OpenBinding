# QFBS HTTP generation

```python
from pathlib import Path
import csv, json, collections
ROOT = Path.cwd()
ART = ROOT / ".artifacts/qacobench"
```

```python
manifest = ROOT / "datasets/07_qfbs/manifest.jsonl"
rows = [json.loads(line) for line in manifest.read_text(encoding="utf-8").splitlines()]
print("Instances:", len(rows))
print("Groups:", dict(collections.Counter(row["group"] for row in rows)))
for group in ("guaranteed", "non_guaranteed"):
    print(group, dict(collections.Counter(row["row"][0] for row in rows if row["group"] == group)))
print("Actual constraint counts:", dict(collections.Counter(row["actual_constraint_count"] for row in rows)))
```
