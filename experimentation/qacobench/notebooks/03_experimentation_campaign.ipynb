```python
from pathlib import Path
import csv, json, collections
ROOT = Path.cwd()
ART = ROOT / ".artifacts/qacobench"
```

```python
results = ART / "results.jsonl"
counts = collections.Counter()
triples = set()
for line in results.open(encoding="utf-8"):
    if not line.strip():
        continue
    row = json.loads(line)
    counts[row["status"]] += 1
    triples.add((row["suite"], row["instance_id"], row["engine"]))
print("Runs:", sum(counts.values()), "Unique triples:", len(triples), "Expected:", 46350)
print("Statuses:", dict(counts))
```
